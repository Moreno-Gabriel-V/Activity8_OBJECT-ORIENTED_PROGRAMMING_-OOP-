```python
#3
# Online Shopping Cart System (FIXED)

products = [
    {"name": "Rice", "price": 50, "stock": 10},
    {"name": "Milk", "price": 60, "stock": 8},
    {"name": "Bread", "price": 40, "stock": 15},
    {"name": "Air", "price": 0, "stock": 9999},
    {"name": "Eggs", "price": 10, "stock": 30}
]

cart = []

def show_products():
    print("\n--- Available Products ---")
    for i, p in enumerate(products):
        print(f"{i+1}. {p['name']} - ₱{p['price']} (Stock: {p['stock']})")

def show_cart():
    print("\n--- Cart ---")
    total = 0

    if len(cart) == 0:
        print("Cart is empty.")
        return 0

    for item in cart:
        print(f"{item['name']} - ₱{item['price']}")
        total += item["price"]

    print(f"Total: ₱{total}")
    return total


while True:
    print("\n--- MENU ---")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. Remove from Cart")
    print("4. View Cart")
    print("5. Checkout")
    print("6. Exit")

    choice = input("Enter choice: ")

    # VIEW PRODUCTS
    if choice == "1":
        show_products()

    # ADD TO CART (FIXED STOCK CHECK)
    elif choice == "2":
        show_products()

        try:
            index = int(input("Select product number: ")) - 1

            if index < 0 or index >= len(products):
                print("Invalid product selection.")
                continue

            if products[index]["stock"] <= 0:
                print("Out of stock.")
                continue

            # reduce stock FIRST (safe handling)
            products[index]["stock"] -= 1

            # add copy to cart
            cart.append({
                "name": products[index]["name"],
                "price": products[index]["price"]
            })

            print("Item added to cart.")

        except ValueError:
            print("Invalid input. Please enter a number.")

    # REMOVE FROM CART (FIXED STOCK RESTORATION)
    elif choice == "3":
        if len(cart) == 0:
            print("Cart is empty.")
            continue

        show_cart()
        name = input("Enter product name to remove: ").strip().lower()

        found = False

        for item in cart:
            if item["name"].lower() == name:
                cart.remove(item)

                # restore stock
                for p in products:
                    if p["name"].lower() == name:
                        p["stock"] += 1
                        break

                print("Item removed from cart.")
                found = True
                break

        if not found:
            print("Item not found in cart.")

    # VIEW CART
    elif choice == "4":
        show_cart()

    # CHECKOUT (FIXED RESET)
    elif choice == "5":
        total = show_cart()

        if total > 0:
            print("Checkout successful. Thank you for shopping!")
            cart.clear()
        else:
            print("Nothing to checkout.")

    # EXIT
    elif choice == "6":
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Try again.")
```

```python
# 2
# Employee Payroll System

employees = []

num_employees = int(input("Enter number of employees: "))

if num_employees < 1:
    print("Error: At least one employee is required.")
    exit()

for i in range(num_employees):
    print(f"\nEmployee {i+1}")
    name = input("Enter name: ").strip()
    hourly_rate = float(input("Enter hourly wage: "))
    hours_worked = float(input("Enter hours worked: "))

    if hourly_rate < 0 or hours_worked < 0:
        print("Invalid input: wages and hours cannot be negative.")
        continue

    overtime_pay = 0
    base_hours = min(hours_worked, 160)
    overtime_hours = max(0, hours_worked - 160)

    base_salary = base_hours * hourly_rate

    if overtime_hours > 0:
        overtime_pay = overtime_hours * hourly_rate * 1.5

    total_salary = base_salary + overtime_pay

    employees.append({
        "name": name,
        "hours": hours_worked,
        "base_salary": base_salary,
        "overtime_pay": overtime_pay,
        "total_salary": total_salary
    })

print("\n--- Payroll Report ---")
for emp in employees:
    print(f"\nName: {emp['name']}")
    print(f"Hours Worked: {emp['hours']}")
    print(f"Base Salary: {emp['base_salary']:.2f}")
    print(f"Overtime Pay: {emp['overtime_pay']:.2f}")
    print(f"Total Salary: {emp['total_salary']:.2f}")
```

```python
#1
# Library Book Management System
library = []

# Add books
num_books = int(input("How many books do you want to add? "))

for i in range(num_books):
    print(f"\nBook {i+1}")
    title = input("Enter title: ").strip()
    author = input("Enter author: ").strip()
    status = input("Enter status (available/borrowed): ").strip().lower()

    if status not in ["available", "borrowed"]:
        print("Invalid status. Setting default to 'available'.")
        status = "available"

    library.append({
        "title": title,
        "author": author,
        "status": status
    })

# Function to display books
def display_books():
    print("\n--- Library Books ---")
    for book in library:
        print(f"{book['title']} by {book['author']} - {book['status']}")

# Borrow book
def borrow_book(title):
    for book in library:
        if book["title"].lower() == title.lower():
            if book["status"] == "borrowed":
                print("Sorry, this book is already borrowed.")
                return
            book["status"] = "borrowed"
            print(f"You have successfully borrowed '{book['title']}'.")
            return
    print("Book not found.")

# Return book
def return_book(title):
    for book in library:
        if book["title"].lower() == title.lower():
            if book["status"] == "available":
                print("This book is not currently borrowed.")
                return
            book["status"] = "available"
            print(f"You have successfully returned '{book['title']}'.")
            return
    print("Book not found.")

# Menu system
while True:
    print("\n--- MENU ---")
    print("1. Display books")
    print("2. Borrow book")
    print("3. Return book")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        display_books()
    elif choice == "2":
        title = input("Enter book title to borrow: ")
        borrow_book(title)
    elif choice == "3":
        title = input("Enter book title to return: ")
        return_book(title)
    elif choice == "4":
        print("Exiting program...")
        break
    else:
        print("Invalid choice. Try again.")
```
