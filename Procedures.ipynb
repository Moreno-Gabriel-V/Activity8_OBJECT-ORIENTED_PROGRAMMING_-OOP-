```python
# 8
from abc import ABC, abstractmethod

class Appliance(ABC):

    @abstractmethod
    def turn_on(self):
        pass

class Refrigerator(Appliance):

    def turn_on(self):
        return "Refrigerator is now cooling."

ref = Refrigerator()

print(ref.turn_on())
```

```python
# 7
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def display(self):
        return f"({self.x}, {self.y})"

vector1 = Vector(5, 3)
vector2 = Vector(2, 1)

add_result = vector1 + vector2
sub_result = vector1 - vector2

print("Addition:", add_result.display())
print("Subtraction:", sub_result.display())
```

```python
# 6
class TemperatureConverter:

    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9/5) + 32

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = TemperatureConverter.celsius_to_fahrenheit(celsius)

print("Temperature in Fahrenheit:", fahrenheit)
```

```python
# 4
class Device:
    def power_on(self):
        return "Device is powering on."

class Smartphone(Device):
    def power_on(self):
        return "Smartphone is booting up."

class Laptop(Device):
    def power_on(self):
        return "Laptop is starting."

def start_device(device):
    print(device.power_on())

phone = Smartphone()
laptop = Laptop()

start_device(phone)
start_device(laptop)
```

```python
# 3
class Device:
    def __init__(self, device_name):
        self.device_name = device_name

    def power_on(self):
        return f"{self.device_name} is powering on."

class Smartphone(Device):
    def power_on(self):
        return f"{self.device_name} smartphone is booting up."

device1 = Device("Generic Device")
phone1 = Smartphone("Samsung")

print(device1.power_on())
print(phone1.power_on())
```

```python
# 2
class StudentRecord:
    def __init__(self):
        self.__grades = []

    def add_grade(self, grade):
        self.__grades.append(grade)

    def get_grades(self):
        return self.__grades

student = StudentRecord()

student.add_grade(90)
student.add_grade(85)
student.add_grade(95)

print(student.get_grades())
```

```python
# 1
class Laptop:
    def __init__(self, brand, processor, ram):
        self.brand = brand
        self.processor = processor
        self.ram = ram

    def get_specs(self):
        return f"Laptop: {self.brand}, Processor: {self.processor}, RAM: {self.ram}"

my_laptop = Laptop("Dell", "Intel i7", "16GB")

print(my_laptop.get_specs())
```
